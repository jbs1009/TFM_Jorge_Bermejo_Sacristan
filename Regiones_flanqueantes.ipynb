```python
import pandas as pd
import requests
import time

from Bio import Entrez, SeqIO
```

```python
df = pd.read_csv("download/t_common_variant.txt", sep = "\t", index_col = False)
```

```python
df_SNPs = df[["Chr", "SNP_position", "effect_allele", "alternate_allele"]]
```

```python
df_SNPs
```

```python
df_SNPs = df_SNPs.dropna()
```

```python
df_SNPs = df_SNPs.reset_index(drop=False)[["Chr", "SNP_position", "effect_allele", "alternate_allele"]]
```

```python
df_SNPs
```

```python
df_copia = pd.DataFrame(columns = [i for i in df_SNPs])
```

```python
df_copia
```

```python
df_SNPs_copia = pd.DataFrame(columns = [i for i in df_SNPs])

for i, fila in enumerate(df_SNPs["Chr"]):
    
    if "rs" in fila:
        continue
        
    else:
        df_SNPs_copia = pd.concat([df_SNPs_copia, df_SNPs.iloc[[i]]])
```
