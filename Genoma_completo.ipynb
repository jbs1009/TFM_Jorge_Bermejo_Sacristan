```python
import gzip
import shutil
```

```python
genoma_comprimido = "Homo_sapiens.GRCh37.dna_sm.primary_assembly.fa.gz"

genoma = "Homo_sapiens.GRCh37.completo.fa"

with gzip.open(genoma_comprimido, 'rb') as file_in:
    with open(genoma, 'wb') as file_out:
        shutil.copyfileobj(file_in, file_out)
```
